```python
import matplotlib.pyplot as plt
import numpy as np

print("Creating test graph...")

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2)
plt.title("Test Graph - It Works!", fontsize=16)
plt.xlabel("X", fontsize=12)
plt.ylabel("sin(X)", fontsize=12)
plt.grid(True, alpha=0.3)

# Save the graph
plt.savefig('../output/graphs/test_graph.png', dpi=300, bbox_inches='tight')
print("✅ Graph saved to output/graphs/test_graph.png")
```
